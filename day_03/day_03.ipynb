What were the outputs of the pipeline?

The outputs of the pipeline are:
- a report: that contain the main reporting output of the workflow (.html file)
- plots: graphical output of the workflow. Inside, there are different images that can be directly used in reports or publications. This images are also included in the HTML report

Finally, there is a methods sections, where they set the parameters that they used for the different processes that are happening insede this pipeline.
