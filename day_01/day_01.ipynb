1. What is nf-core?

nf-core is an open-source community that develops and maintains pipelines using Nextflow. It provides standardised guidelines and templates for building and sharing bioinformatics pipelines. These make analyses easier to run, reproduce and share across different environments.

2. How many pipelines are there currently in nf-core?

There are a total of 157 pipelines currently available in nf-core. 

3. Are there any non-bioinformatic pipelines in nf-core?

Yes, there are some non-bioinformatic pipelines in nf-core. Some examples I found are:
- **rangeland**: a geografical analysis pipeline for remotely sensed satellite imagery and land-cover changes.
- **provenancereport**: a reporting pipeline that validates a samplesheet and generates reproducible Quarto reports. 
- **datasync**: a Nextflow pipeline for copying files and directories between storage locations while documenting their integrity. 
- **meerpipe**: an astronomy-related pipeline.
- **spinningjenny**: a computational pipeline that simulates the First Industrial Revolution using Agent Based Models.

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A **pipeline** is a sequence of connected steps or tasks used to process data. The output of one step is usually used as the input for the next one. Pipelines are used to automate analyses, make them reproducible and reduce manual work and errors. 

In bioinformatics, a pipeline can also be defined as an automated sequence of computational tools and algorithms used to process and analyse complex biological data. 

5. Why do you think nf-core adheres to strict guidelines?

I think nf-core follows such strict guidelines to make pipelines consistent and reproducible. Having the same standards also makes it easier to understand, test, share and maintain. It can also help to reduce errors. 

6. What are the main features of nf-core pipelines?

The main features of nf-core pipelines are that they are standardised, reproducible, portable and well documented. They are also tested and modular and reusable components. In addition, they are open source and community maintained, which makes them easier to share, improve and run across different computing environments.
