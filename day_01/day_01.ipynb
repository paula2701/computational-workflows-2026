**1. What is nf-core?**

nf-core is an open-source community that develops and maintains pipelines using Nextflow. It provides standardised guidelines and templates for building and sharing bioinformatics pipelines. These make analyses easier to run, reproduce and share across different environments.

**2. How many pipelines are there currently in nf-core?**

There are a total of 157 pipelines currently available in nf-core. 

**3. Are there any non-bioinformatic pipelines in nf-core?**

Yes, there are some non-bioinformatic pipelines in nf-core. Some examples I found are:
- **rangeland**: a geografical analysis pipeline for remotely sensed satellite imagery and land-cover changes.
- **provenancereport**: a reporting pipeline that validates a samplesheet and generates reproducible Quarto reports. 
- **datasync**: a Nextflow pipeline for copying files and directories between storage locations while documenting their integrity. 
- **meerpipe**: an astronomy-related pipeline.
- **spinningjenny**: a computational pipeline that simulates the First Industrial Revolution using Agent Based Models.

**4. Let's go back a couple of steps. What is a pipeline and what do we use it for?**

A **pipeline** is a sequence of connected steps or tasks used to process data. The output of one step is usually used as the input for the next one. Pipelines help us to automate analyses, make them reproducible and reduce manual work and errors. 

In bioinformatics, a pipeline can also be defined as an automated sequence of computational tools and algorithms used to process and analyse complex biological data. 

**5. Why do you think nf-core adheres to strict guidelines?**

I think nf-core follows such strict guidelines to make pipelines consistent, reproducible, and easier to mantain. Using the standards also makes pipelines easier to understant, test,review, and share between different users and developers. In addition, following clear guidelines can help reduce errors and make it easier to update or modify a pipeline without changing its overall structure. 

**6. What are the main features of nf-core pipelines?**

The main features of nf-core pipelines are that they are standardised, reproducible, portable and well documented. They are also tested and build using modular and reusable components. In addition, they are open source and community maintained, which makes them easier to share, improve and run across different computing environments.

**Check out the current directory. Next to the outdir you specified, what else has changed?**

Besides the folder test_results1/, Nextflow created or updated different files and folders:
- work/: contains the files generated while the tasks are running, such as scripts, logs, and intermediate files. When the pipeline is run again without -resume, new task folders can be created. 
- .nextflow/: stores information about previous runs and the task cache, which Nextflow can use when resuming a pipeline. 
- .nextflow.log: contains the log of the most recent run. Older logs are kept as .nextflow.log.1, .nextflow.log.2, and so on.

When -resume is used, Nextflow can reuse the results from previous task executions instead of running those tasks again. 

```python
# delete the work directory and run the pipeline again using -resume. What did change?
!nextflow run nf-core/differentialabundance -profile test,docker --outdir test_results1 -r 2.0.0 -c /home/paula/.ssh/computational-workflows-2026/nextflow.config -resume
```

**What changed?**

After deleting the work/ directory, I ran the pipeline again using -resume. However, Nextflow could not reuse the previous task results, so it had to run the tasks again and recreated the work/ directory.

The information about the previous run was still stored in .nextflow/, so the run history and cache were not completely lost. However, the actual files produced by the tasks were no longer available because they had been deleted together with the work/ directory. Because of this, Nextflow did not have the files it needed to continue from the previous execution.

I also noticed that this run took longer than the previous one with -resume, which made sense because the tasks had to be executed again instead of being reused. A new set of execution reports was also added to test_results1/pipeline_info/ for this run.

From this, I understood that -resume does not only depend on the cache information stored in .nextflow/. The corresponding task files in work/ also need to be present so that Nextflow can reuse the previous results.

### **What is differential abundance analysis?**

Differential abundance analysis helps us compare groups of biological samples to identify features (such as genes or proteins), whose abundance changes significantly between the groups. As a result of this analysis, several plots and tables are generated, which help us visualize and interpret these differences. 

**Give the most important plots from the report:**

```python
from IPython.display import Image, display

display(Image(
    filename = "/home/paula/.ssh/computational-workflows-2026/day_01/test_results1/plots/differential/rnaseq_deseq2_gsea/treatment_mCherry_hND6__SRP254919/png/volcano.png",
    width=900
))
```

The volcano plot is one of the plots that nf-core/differentialabundance generated that I found important for further analysis. This plot summarises differences in gene expression between hND6 and mCherry. Each dot represents one gene. On the x-axis, the log2 fold change shows the direction and size of the expression change. Positive values mean that the gene is more highly expressed in hND6, while negative values mean higher expression in mCherry. A value around zero means that there is little or no difference between the two groups. For example, a log2 fold change of 1 means that the gene is expressed about twice as much in one group compared with the other.

The y-axis represents the -log10 adjusted p-value. Genes that appear higher on the plot have smaller adjusted p-values, which means there is stronger statistical evidence that their expression is different between the two groups. The p-values are adjusted because many genes are tested at the same time. 

In general, the genes that we are most interested in, are the ones that have both a substantial fold change and a sufficiently small adjusted p-value. In the plot that nf-core/differentialabundance generated, the orange points are the ones that meet the plots differential-expression criteria.

Another plot that is important for the differential abundance analysis is the box plot. This plot shows the expression levels of several selected genes in the two experimental conditions (hND6 and mCherry). Each panel represents a different gene. This is useful because we can not only see whether the expression differs between the groups, but also how much variation there is between the individual samples.

For genes such as Igha, Ighg2b, Igkc and Jchain there is a very clear separation between hND6 and mCherry. Their adjusted p-values are displayed as 0 in the plot, which most likely means that they are extremely small and have been rounded to 0. This provides very strong statistical evidence that the expression of these genes differs between the two conditions. However, the adjusted p-value tells us about the statistical evidence for a difference, rather than the size of the difference itself.

The genes Iglc1 and Pde6g show a slightly different situation. Their expression levels also appear to differ between hND6 and mCherry, but there is more variability between the individual samples. Their adjusted p-values are 0.008 and 0.024, respectively. Both values are still below the commonly used threshold of 0.05 and are therefore considered statistically significant, although the statistical evidence is not as strong as for the genes with extremely small adjusted p-values.

The principal component analysis (PCA) plot gives us an overall view of how similar or different the samples are. In this analysis, the 500 most variable genes are used to summarise the main differences between them.

Each point in the plot represents one sample. The colours indicate whether the sample belongs to the hND6 or mCherry condition. The position of the points is important since samples that are close together have similar gene expression patterns, while samples that are further apart are more different. 

The x-axis represents PC1, which explains 45.3% of the variation in the data, while PC2 explains another 22.1% (y-axis). Both components account for 67.4% of the total variation, so they capture a large part of the differences between the samples.

In this plot, the two groups show some separation, especially along PC2, altough there is still considerable variation within each group. This suggests that the experimental condition may explain part of the cariation between the samples, but probably not all of it. PCA is therefore useful for checking whether samples tend to group according to their conition and for identifying samples that behave differently from the rest. However, it does not tell us whether the differences are statistically significant or which specific genes are responsible for them.
